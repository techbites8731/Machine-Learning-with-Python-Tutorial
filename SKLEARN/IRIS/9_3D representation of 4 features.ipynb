```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

iris = load_iris()
```

```python
X = []                 # empty list

for iclass in range(3):                     # 0, 1, 2
    X.append([[], [], []])                 # [[[], [], []]]
    for i in range(len(iris.data)) :                      # 150
        if iris.target[i]==iclass:
            X[iclass][0].append(iris.data[i][0])
            X[iclass][1].append(iris.data[i][1])
            X[iclass][2].append(sum(iris.data[i][2:]))    # combined feature 2 & 3

X            # # [[[], [], []], [[], [], []],[[], [], []]]      -- shape(3,3,0)   - 3 rows, each row has 3 columns and 1 row
```

```python
rows = len(X)
columns = len(X[0])                  # this counts the items inside the first list
rows, columns
```

```python
colors = ['r', 'g', 'y']
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

for iclass in range(3):
    ax.scatter(X[iclass][0], X[iclass][1], X[iclass][2], color=colors[iclass])

plt.show()
```
